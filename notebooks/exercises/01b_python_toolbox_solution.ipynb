```python
# === 題目 1：import 與統計 ===
import statistics as stats

hospital_days = [5, 3, 12, 7, 4, 9, 6, 8]

mean_days = stats.mean(hospital_days)
median_days = stats.median(hospital_days)

print(f"平均住院天數: {mean_days:.1f}")
print(f"住院天數中位數: {median_days:.1f}")
```

```python
# === 題目 2：型別轉換 ===
cases_str = "121"
population_str = "280"
deaths_str = "19"

cases = int(cases_str)
population = int(population_str)
deaths = int(deaths_str)

attack_rate = cases / population
cfr = deaths / cases

print(f"侵襲率: {attack_rate:.2%}")
print(f"致死率: {cfr:.2%}")
```

```python
# === 題目 3：讀懂錯誤 ===

# Bug 1: NameError — 變數名少了 s
total_residents = 280
print(f"Bug 1 修正: total_residents = {total_residents}")

# Bug 2: TypeError — 字串不能和數字比較
rate = float("0.432")  # 先轉成 float
threshold = rate > 0.5
print(f"Bug 2 修正: {rate} > 0.5 → {threshold}")

# Bug 3: KeyError — key 是小寫 "wing" 不是 "Wing"
patient = {"age": 72, "sex": "M", "floor": 3, "wing": "B"}
print(f"Bug 3 修正: wing = {patient['wing']}")
```

```python
# === 題目 4：try/except ===
wing_data = ["15", "10", "N/A", "25", "missing", "27"]

clean_data = []
skipped = 0
for val in wing_data:
    try:
        clean_data.append(int(val))
    except ValueError:
        skipped += 1

print(f"成功: {len(clean_data)} 筆")
print(f"跳過: {skipped} 筆")
print(f"總和: {sum(clean_data)}")
```

```python
# === 題目 5：字串清理 + 迴圈 ===
exposures_raw = [" Shower ", "shower", "HYDROTHERAPY", " hydrotherapy", "Shower", "unknown"]

cleaned = []
for raw in exposures_raw:
    cleaned.append(raw.strip().lower())

shower_count = cleaned.count("shower")
hydro_count = cleaned.count("hydrotherapy")

print(f"清理後: {cleaned}")
print(f"shower 出現 {shower_count} 次")
print(f"hydrotherapy 出現 {hydro_count} 次")
```

```python
# === 題目 6（挑戰題）：綜合應用 ===
patients = [
    {"id": "P001", "floor": "3", "infected": "1"},
    {"id": "P002", "floor": "2", "infected": "0"},
    {"id": "P003", "floor": "3", "infected": "1"},
    {"id": "P004", "floor": "1", "infected": "N/A"},
    {"id": "P005", "floor": "2", "infected": "1"},
    {"id": "P006", "floor": "3", "infected": "0"},
    {"id": "P007", "floor": "1", "infected": "1"},
    {"id": "P008", "floor": "2", "infected": "1"},
]

# 統計字典：{樓層: {"total": N, "infected": N}}
floor_stats = {}

for p in patients:
    floor = p["floor"]
    if floor not in floor_stats:
        floor_stats[floor] = {"total": 0, "infected": 0}
    
    try:
        infected = int(p["infected"])
        floor_stats[floor]["total"] += 1
        floor_stats[floor]["infected"] += infected
    except ValueError:
        # N/A — 跳過此筆但不計入分母
        pass

print("各樓層侵襲率：")
for floor in sorted(floor_stats):
    s = floor_stats[floor]
    ar = s["infected"] / s["total"] if s["total"] > 0 else 0
    print(f"  {floor}F: {ar:.1%} ({s['infected']}/{s['total']})")
```

```python
# === 題目 7：statistics 模組 ===
import statistics
onset_hours = [24, 30, 28, 36, 26, 40, 32, 29, 34, 27]
print(f"平均 = {statistics.mean(onset_hours):.1f} 小時")
print(f"中位數 = {statistics.median(onset_hours):.1f} 小時")
print(f"標準差 = {statistics.stdev(onset_hours):.1f} 小時")
```

```python
# === 題目 8：try/except 與型別轉換 ===
raw = ["12", "8", "N/A", "15", "", "9", "twenty", "20"]
total, valid, bad = 0, 0, 0
for x in raw:
    try:
        total += int(x); valid += 1
    except ValueError:
        bad += 1
print(f"有效天數 = {valid}，總病例數 = {total}，無法解析 = {bad} 筆")
```

```python
# === 題目 9（挑戰題）：定義函式 + 錯誤處理 ===
def cfr(deaths, cases):
    """致死率 = 死亡數 / 病例數；cases <= 0 時丟出 ValueError。"""
    if cases <= 0:
        raise ValueError("cases 必須大於 0")
    return deaths / cases

for d, c in [(12, 180), (5, 0)]:
    try:
        print(f"cfr({d}, {c}) = {cfr(d, c):.1%}")
    except ValueError as e:
        print(f"cfr({d}, {c}) 發生錯誤：{e}")
```
