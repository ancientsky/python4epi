```python
# === 題目 1：計算 3 樓 B 翼的基本指標 ===
residents_3b = 47
infected_3b = 27
deaths_3b = 5
hospitalized_3b = 18

attack_rate_3b = infected_3b / residents_3b
cfr_3b = deaths_3b / infected_3b
hosp_rate_3b = hospitalized_3b / infected_3b

print(f"3B 翼侵襲率 (Attack rate): {attack_rate_3b:.2%}")
print(f"3B 翼致死率 (CFR): {cfr_3b:.2%}")
print(f"3B 翼住院率: {hosp_rate_3b:.2%}")
```

```python
# === 題目 2：條件判斷 ===
if attack_rate_3b > 0.50:
    print("建議立即關閉該翼區淋浴設備")
else:
    print("持續監測中")
```

```python
# === 題目 3（挑戰題）：用字典 + 迴圈比較各翼區 ===
wings = {
    "1A": {"residents": 44, "infected": 15},
    "1B": {"residents": 47, "infected": 10},
    "2A": {"residents": 44, "infected": 24},
    "2B": {"residents": 50, "infected": 25},
    "3A": {"residents": 48, "infected": 20},
    "3B": {"residents": 47, "infected": 27},
}

max_ar = 0
max_wing = ""

for name, data in wings.items():
    ar = data["infected"] / data["residents"]
    print(f"{name}: 侵襲率 {ar:.1%}  ({data['infected']}/{data['residents']})")
    if ar > max_ar:
        max_ar = ar
        max_wing = name

print(f"\n侵襲率最高的翼區：{max_wing}（{max_ar:.1%}）")
```
