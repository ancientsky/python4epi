```python
import pandas as pd\nimport matplotlib.pyplot as plt\nimport seaborn as sns\nimport plotly.express as px
```

```python
raw = pd.read_csv('data/synthetic/line_list.csv')\nprint(raw.head())\nprint(raw.columns.tolist())
```

```python
raw['date_onset'] = pd.to_datetime(raw['date_onset'], errors='coerce')\nraw['epi_week'] = raw['date_onset'].dt.isocalendar().week\nweekly = raw.groupby('epi_week').size().rename('cases').reset_index()\ndaily = raw.groupby('date_onset').size().rename('cases')\nweekly
```

```python
plt.figure(figsize=(9, 3.5))\nplt.bar(daily.index.astype(str), daily.values)\nplt.title('Epidemic Curve (By Onset Date)')\nplt.xlabel('Onset Date')\nplt.ylabel('Cases')\nplt.xticks(rotation=45)\nplt.tight_layout()\nplt.show()
```

```python
by_loc = raw.groupby('location').size().rename('cases').reset_index()\nplt.figure(figsize=(6, 3.5))\nsns.barplot(data=by_loc, x='location', y='cases', hue='location', legend=False)\nplt.title('Cases by Location')\nplt.tight_layout()\nplt.show()
```

```python
fig = px.line(weekly, x='epi_week', y='cases', markers=True, title='Interactive Weekly Trend')\nfig.show()
```
