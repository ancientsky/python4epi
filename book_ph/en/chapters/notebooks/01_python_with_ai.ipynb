## 1. Values and variables

A **variable** is a name for a value — like writing a label on a folder. The `=` sign
means *"put this value in that folder"*, not *"is equal to"*.

- `7` is a whole number (an *integer*).
- `"Ilog"` is text (a *string*): text always goes inside quotes.
- `True` / `False` are yes/no values (*booleans*).

`print()` shows a value on the screen. Text after `#` is a **comment**: Python ignores it;
it is a note for humans. An **f-string** puts values inside a sentence: write `f` before
the quotes and the variable name inside `{ }` (last line below).

````python
forms_on_desk = 6            # how many paper forms Maya has this morning
barangay = "Ilog"            # text goes inside quotes
flood_reached_house = True   # a yes/no value

print(forms_on_desk)
print(barangay)
print(flood_reached_house)
print(f"Maya has {forms_on_desk} forms from Barangay {barangay}.")
````

## 2. Lists: many values in one variable

A **list** keeps values in order, inside square brackets `[ ]`. Python counts positions
from **0**, so the *first* item is `ages[0]` — the most common surprise for beginners
(and a common AI slip too).

`len()` counts the items; `sum()` adds them up.

````python
ages = [34, 9, 41, 15, 52, 23]   # ages on the six forms (made up)

print(len(ages))     # how many items?
print(ages[0])       # the FIRST item is position 0
print(ages[-1])      # -1 means "the last item"
print(sum(ages) / len(ages))   # the mean age
````

## 3. Dictionaries: one form, many fields

A **dictionary** stores *labelled* values: a **key** (the label) and a **value**,
inside curly braces `{ }`. One notification form fits nicely in one dictionary.
We use a letter, not a name, to tell the forms apart — a good privacy habit from day one.

Maya's whole stack is then a **list of dictionaries**, one per form. Look at the six forms
closely: you will count them by hand in a minute.

````python
form_a = {"form": "A", "barangay": "Ilog", "disease": "leptospirosis", "age": 34, "waded": True}

print(form_a["disease"])   # read one value by its key
print(form_a["age"])

# Maya's whole stack: a LIST of dictionaries, one per form (all invented)
forms = [
    {"form": "A", "barangay": "Ilog",   "disease": "leptospirosis", "age": 34, "waded": True},
    {"form": "B", "barangay": "Ilog",   "disease": "dengue",        "age": 9,  "waded": False},
    {"form": "C", "barangay": "Wawa",   "disease": "leptospirosis", "age": 41, "waded": True},
    {"form": "D", "barangay": "Sapa",   "disease": "dengue",        "age": 15, "waded": False},
    {"form": "E", "barangay": "Ilog",   "disease": "leptospirosis", "age": 52, "waded": False},
    {"form": "F", "barangay": "Talaba", "disease": "dengue",        "age": 23, "waded": True},
]
print(f"{len(forms)} forms")
````

## 4. A loop and an `if`: counting like a tally sheet

Maya asked her AI assistant, Tala:

> *I have a Python list called `forms`. Each item is a dictionary with keys `form`,
> `barangay`, `disease`, `age`, `waded` (True/False). Two made-up items:
> `{"form": "A", "barangay": "Ilog", "disease": "leptospirosis", "age": 34, "waded": True}` and
> `{"form": "B", "barangay": "Ilog", "disease": "dengue", "age": 9, "waded": False}`.
> Write beginner-friendly code that counts the leptospirosis forms. Explain each line.*

**Count by hand first.** Run your finger down the list: leptospirosis forms are A, C
and E — **3**. Write that down *before* you run anything.

Tala's answer is below. Read it like a tally sheet: start at zero; look at each form;
**if** it is leptospirosis, add one. The indented lines (4 spaces) belong to the loop
or to the `if`. The last line is our **check**: `assert` stops with an error if the
statement is false and stays silent if it is true — silence is good news.

````python
lepto_count = 0                      # start the tally at zero
for f in forms:                      # look at each form, one at a time
    if f["disease"] == "leptospirosis":   # == asks "is it equal?"
        lepto_count = lepto_count + 1     # add one tally mark

print(f"Leptospirosis forms: {lepto_count}")
assert lepto_count == 3, "Tally does not match the hand count!"   # our check
````

## 5. Tala got it wrong: the denominator

Maya's next question: *"What percentage of the **leptospirosis** forms waded in
floodwater?"* Tala sent back the code in the next cell. It runs without any error.
Before you run it, count by hand: how many leptospirosis forms are there, and how many
of them waded? (`and` means both tests must be true.)

````python
# Tala's version, run as-is (it has a mistake -- can you see it before reading on?)
waded_lepto = 0
for f in forms:
    if f["disease"] == "leptospirosis" and f["waded"]:
        waded_lepto = waded_lepto + 1
percent_waded = waded_lepto / len(forms) * 100
print(f"Tala's answer: {percent_waded:.1f}%")
````

**Count by hand.** Leptospirosis forms: A, C, E (3). Of those, waded: A and C (2).
So the answer is 2 out of 3 — about **66.7 %**, not 33.3 %.

The **numerator** (2) was right. The **denominator** was wrong: `len(forms)` counts *all
six* forms, dengue included. A percentage is only as good as its denominator — always
ask *"percent of what?"*

The fix: write a small **function** — a reusable recipe with a name. `def` starts it,
the values in brackets are its inputs, and `return` hands back the result.

The line `[f for f in forms if f["disease"] == "leptospirosis"]` is a one-line loop: it
makes a new list that keeps only the forms passing the test. Read it as *"each form f,
from forms, if it is leptospirosis"*. The
`: int` and `-> float` labels are optional notes saying what kind of value goes in and
comes out. The function also refuses to divide by zero with a clear message — and the
end of the cell **checks** both the hand count and that edge case.

````python
def percent(part: int, whole: int) -> float:
    # Return part as a percentage of whole.
    if whole == 0:
        raise ValueError("The denominator is 0: there is nothing to take a percentage of.")
    return part / whole * 100


lepto_forms = [f for f in forms if f["disease"] == "leptospirosis"]   # only the lepto forms
waded_lepto = len([f for f in lepto_forms if f["waded"]])

answer = percent(waded_lepto, len(lepto_forms))
print(f"{waded_lepto} of {len(lepto_forms)} leptospirosis forms waded: {answer:.1f}%")

# Check 1: matches the hand count.  Check 2: the edge case of a zero denominator.
assert round(answer, 1) == 66.7
try:
    percent(1, 0)
except ValueError as error:
    print("Good, the function refuses:", error)
````

## 6. From six forms to the whole line list: pandas

Counting six forms with a loop is fine. For hundreds of rows we use **pandas**, the
spreadsheet of Python. `import pandas as pd` loads it and gives it the short name `pd`.

We open the **clean, de-identified** Santa Luz line list (no names, phones or
addresses — chapter 02 shows how it got that way). The path is relative to the
repository folder. `.shape` gives (rows, columns); `df.columns` lists the column names —
copy them **exactly** when you ask Tala for code (`date_onset`, not "onset date");
`.head()` shows the first five rows.

````python
import pandas as pd

df = pd.read_csv("data/synthetic/ph/santa_luz_linelist.csv")
print(df.shape)
print(list(df.columns))
df.head()
````

`value_counts()` does what our loop did — a tally — for a whole column at once. Then we
**check** it: the counts must add up to the number of rows.

Then a first chart: cases by **band** of barangays (riverside, coastal, upland) and
disease. `pd.crosstab` makes a small table of counts; `.plot.barh` draws it as
horizontal bars.

````python
counts = df["disease"].value_counts()
print(counts)

assert counts.sum() == len(df), "Some rows were not counted!"
print(f"Total: {counts.sum()} cases = {len(df)} rows")

import matplotlib.pyplot as plt

table = pd.crosstab(df["band"], df["disease"]).loc[["upland", "coastal", "riverside"]]
print(table)

ax = table.plot.barh(color=["#CE1126", "#0038A8"], figsize=(7, 3.5))
ax.set_xlabel("Number of cases (reported by 30 Sep 2026)")
ax.set_ylabel("")
ax.set_title("Santa Luz City (fictional): cases by band and disease")
ax.legend(title="")
plt.tight_layout()
plt.show()
````

## 7. Reading an error calmly

Errors are messages, not insults. Python reads your code, hits a problem and tells you
**what** went wrong (the error *type*) and **where** (the last line it tried). Read the
**last line first**. A typical one:

```text
KeyError: 'Disease'
```

`KeyError` means *"I looked for this key (column name) and it is not there"*. The
column is called `disease`, small *d*: Python is case-sensitive.

The cell below makes that mistake on purpose. The `try:` / `except` lines catch the
error and print it, so the notebook keeps running.

**Asking Tala to explain a line.** When a line confuses you, paste **that line only**
(code, never data) and ask in plain words. For example:

> print for a tiny made-up table with 3 rows: `df["disease"].value_counts()`.*

Then check the explanation against what the line really prints.

````python
try:
    df["Disease"]                 # capital D: no such column
except KeyError as error:
    print("KeyError:", error)
    print("Columns that do exist start with:", list(df.columns)[:4])
````
