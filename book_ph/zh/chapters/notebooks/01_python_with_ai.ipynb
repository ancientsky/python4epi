## 1. 值與變數

**變數**是一個值的名字，就像在資料夾上寫標籤。`=` 的意思是「*把這個值放進那個資料夾*」，而不是「*等於*」。

- `7` 是整數（*integer*）。
- `"Ilog"` 是文字（*string*）：文字一定要放在引號裡。（Ilog 是一個 barangay（描籠涯，菲律賓最基層的行政區，類似台灣的村里）的名字。）
- `True` / `False` 是「是／否」的值（*boolean*）。

`print()` 會把一個值顯示在畫面上。`#` 之後的文字是**註解**：Python 會忽略它，它是寫給人看的備忘。**f-string** 把值放進句子裡：在引號前面寫 `f`，把變數名稱放在 `{ }` 裡面（見下面最後一行）。

````python
forms_on_desk = 6            # 今天早上 Maya 桌上有幾張紙本通報單
barangay = "Ilog"            # 文字要放在引號裡
flood_reached_house = True   # 一個「是／否」的值

print(forms_on_desk)
print(barangay)
print(flood_reached_house)
print(f"Maya 今天有 {forms_on_desk} 張來自 Barangay {barangay} 的通報單。")
````

## 2. 串列：一個變數裡放很多值

**串列**（list）把值依序放在方括號 `[ ]` 裡。Python 從 **0** 開始數位置，所以*第一個*項目是 `ages[0]`，這是初學者最常感到意外的地方（也是 AI 常犯的小失誤）。

`len()` 計算項目的數量；`sum()` 把它們加起來。

````python
ages = [34, 9, 41, 15, 52, 23]   # 六張單子上的年齡（虛構）

print(len(ages))     # 有幾個項目？
print(ages[0])       # 第一個項目的位置是 0
print(ages[-1])      # -1 代表「最後一個項目」
print(sum(ages) / len(ages))   # 平均年齡
````

## 3. 字典：一張表單，許多欄位

**字典**（dictionary）儲存的是有*標籤*的值：一個**鍵**（標籤）和一個**值**，放在大括號 `{ }` 裡。一張通報單剛好可以放進一個字典。我們用字母而不是姓名來區分每張單子，這是從第一天就該養成的隱私好習慣。

於是 Maya 的整疊單子就是一個**由字典組成的串列**，每張單子一個。仔細看這六張單子：你等一下要用手數它們。

````python
form_a = {"form": "A", "barangay": "Ilog", "disease": "leptospirosis", "age": 34, "waded": True}

print(form_a["disease"])   # 用鍵讀出一個值
print(form_a["age"])

# Maya 的整疊單子：一個由字典組成的串列，每張單子一個（全是虛構的）
forms = [
    {"form": "A", "barangay": "Ilog",   "disease": "leptospirosis", "age": 34, "waded": True},
    {"form": "B", "barangay": "Ilog",   "disease": "dengue",        "age": 9,  "waded": False},
    {"form": "C", "barangay": "Wawa",   "disease": "leptospirosis", "age": 41, "waded": True},
    {"form": "D", "barangay": "Sapa",   "disease": "dengue",        "age": 15, "waded": False},
    {"form": "E", "barangay": "Ilog",   "disease": "leptospirosis", "age": 52, "waded": False},
    {"form": "F", "barangay": "Talaba", "disease": "dengue",        "age": 23, "waded": True},
]
print(f"{len(forms)} 張單子")
````

## 4. 迴圈和 `if`：像劃記表一樣計數

Maya 問她的 AI 助理 Tala：

> *我有一個叫 `forms` 的 Python 串列。每一項是一個字典，鍵有 `form`、`barangay`、`disease`、`age`、`waded`（True/False）。兩個虛構的項目：
> `{"form": "A", "barangay": "Ilog", "disease": "leptospirosis", "age": 34, "waded": True}` 和
> `{"form": "B", "barangay": "Ilog", "disease": "dengue", "age": 9, "waded": False}`。
> 請寫適合初學者的程式碼，計算鉤端螺旋體病（leptospirosis）的單子有幾張，並逐行說明。*

**先用手數。** 用手指順著串列往下看：鉤端螺旋體病的單子是 A、C 和 E，共 **3** 張。在執行任何東西*之前*，先把這個數字寫下來。

Tala 的回答在下面。像讀劃記表一樣讀它：從零開始；一張一張看；**如果**是鉤端螺旋體病，就加一。縮排的那幾行（4 個空格）屬於迴圈或 `if`。最後一行是我們的**檢查**：`assert` 在敘述為假時會報錯停下，為真時保持沉默，沉默就是好消息。

````python
lepto_count = 0                      # 從零開始劃記
for f in forms:                      # 一張一張地看每張單子
    if f["disease"] == "leptospirosis":   # == 問的是「是否相等？」
        lepto_count = lepto_count + 1     # 多劃一筆

print(f"鉤端螺旋體病的單子：{lepto_count}")
assert lepto_count == 3, "劃記的結果和手算的不一致！"   # 我們的檢查
````

## 5. Tala 搞錯了：分母

Maya 的下一個問題：*「**鉤端螺旋體病**的單子中，有百分之幾涉過洪水？」* Tala 回傳了下一個儲存格裡的程式碼。它執行時沒有任何錯誤。執行之前，先用手數：鉤端螺旋體病的單子有幾張？其中有幾張涉過水？（`and` 的意思是兩個條件都必須為真。）

````python
# Tala 的版本，原封不動地執行（它有一個錯誤，你能在往下讀之前先看出來嗎？）
waded_lepto = 0
for f in forms:
    if f["disease"] == "leptospirosis" and f["waded"]:
        waded_lepto = waded_lepto + 1
percent_waded = waded_lepto / len(forms) * 100
print(f"Tala 的答案：{percent_waded:.1f}%")
````

**用手數。** 鉤端螺旋體病的單子：A、C、E（3 張）。其中涉過水的：A 和 C（2 張）。所以答案是三張裡有兩張，大約 **66.7 %**，不是 33.3 %。

**分子**（2）是對的。**分母**是錯的：`len(forms)` 數的是*全部六張*單子，把登革熱也算進去了。百分比的好壞，取決於它的分母，永遠要問「*百分比的分母是誰？*」

修正的方法：寫一個小**函式**，也就是一個有名字、可重複使用的食譜。`def` 開頭，括號裡的值是它的輸入，`return` 把結果交回來。

`[f for f in forms if f["disease"] == "leptospirosis"]` 這一行是一行版的迴圈：它做出一個新串列，只留下通過條件的單子。把它讀成「*forms 裡的每張單子 f，如果它是鉤端螺旋體病*」。`: int` 和 `-> float` 這兩個標註是選擇性的備忘，說明進去的是哪一類值、出來的是哪一類值。這個函式也會在除以零時，用一段清楚的訊息拒絕執行，而儲存格的最後會**檢查**手算結果以及這個邊界情況。

````python
def percent(part: int, whole: int) -> float:
    # 回傳 part 占 whole 的百分比。
    if whole == 0:
        raise ValueError("分母是 0：沒有東西可以算百分比。")
    return part / whole * 100


lepto_forms = [f for f in forms if f["disease"] == "leptospirosis"]   # 只留下鉤端螺旋體病的單子
waded_lepto = len([f for f in lepto_forms if f["waded"]])

answer = percent(waded_lepto, len(lepto_forms))
print(f"{waded_lepto} 張涉過水（鉤端螺旋體病單子共 {len(lepto_forms)} 張）：{answer:.1f}%")

# 檢查 1：和手算的結果一致。  檢查 2：分母為零的邊界情況。
assert round(answer, 1) == 66.7
try:
    percent(1, 0)
except ValueError as error:
    print("很好，函式拒絕了：", error)
````

## 6. 從六張單子到整份 line list：pandas

用迴圈數六張單子沒問題。面對幾百列資料，我們用 **pandas**，也就是 Python 的試算表。`import pandas as pd` 把它載入，並給它一個簡短的名字 `pd`。

我們打開**已清理、已去識別化**的 Santa Luz line list（造冊；沒有姓名、電話或住址，第 02 章說明了它是怎麼變成這樣的）。路徑是相對於程式碼儲存庫資料夾的。`.shape` 給出（列數, 欄數）；`df.columns` 列出欄位名稱，向 Tala 要程式碼時要**完全照抄**（是 `date_onset`，不是「onset date」）；`.head()` 顯示前五列。

````python
import pandas as pd

df = pd.read_csv("data/synthetic/ph/santa_luz_linelist.csv")
print(df.shape)
print(list(df.columns))
df.head()
````

`value_counts()` 做的事，和我們的迴圈一樣，是劃記，只是一次對整個欄位做。然後我們**檢查**它：各項計數加起來必須等於列數。

接著是第一張圖：依 barangay 的**地帶**（河岸、沿海、山區）與疾病來看的病例數。`pd.crosstab` 做出一張小小的計數表；`.plot.barh` 把它畫成橫條圖。

````python
counts = df["disease"].value_counts()
print(counts)

assert counts.sum() == len(df), "有些列沒有被計入！"
print(f"總計：{counts.sum()} 個病例 = {len(df)} 列")

import matplotlib.pyplot as plt
# BEGIN-CJK (中文字型設定：避免標籤顯示為方框)
import pathlib
import matplotlib.font_manager as fm
for _font_dir in map(pathlib.Path, ["/usr/share/fonts", "/usr/local/share/fonts"]):
    if _font_dir.exists():
        for _fp in sorted(_font_dir.rglob("*")):
            if _fp.suffix.lower() in {".ttf", ".ttc", ".otf"} and (
                "CJK" in _fp.name or "WenQuanYi" in _fp.name or "wqy" in _fp.name
            ):
                try:
                    fm.fontManager.addfont(str(_fp))
                except Exception:
                    pass
_discovered = []
for _entry in fm.fontManager.ttflist:
    if any(_kw in _entry.name.lower() for _kw in ("cjk", "wenquanyi", "wqy")):
        if _entry.name not in _discovered:
            _discovered.append(_entry.name)
_preferred = ["Noto Sans CJK TC", "Noto Sans CJK SC", "Noto Sans CJK JP", "Noto Sans TC",
              "Microsoft JhengHei", "WenQuanYi Zen Hei", "SimHei", "Arial Unicode MS",
              "Heiti TC", "DejaVu Sans"]
_cjk_fonts = _discovered + [n for n in _preferred if n not in _discovered]
plt.rcParams["font.sans-serif"] = _cjk_fonts + [
    f for f in plt.rcParams.get("font.sans-serif", []) if f not in _cjk_fonts
]
plt.rcParams["axes.unicode_minus"] = False
del _discovered, _preferred, _cjk_fonts
# END-CJK

table = pd.crosstab(df["band"], df["disease"]).loc[["upland", "coastal", "riverside"]]
print(table)

ax = table.plot.barh(color=["#CE1126", "#0038A8"], figsize=(7, 3.5))
ax.set_xlabel("病例數（截至 2026 年 9 月 30 日通報）")
ax.set_ylabel("")
ax.set_title("Santa Luz 市（虛構）：依地帶與疾病的病例數")
ax.legend(title="")
plt.tight_layout()
plt.show()
````

## 7. 冷靜地讀懂錯誤訊息

錯誤是訊息，不是羞辱。Python 讀你的程式碼，遇到問題，然後告訴你**什麼**出了錯（錯誤*類型*）以及**在哪裡**（它最後嘗試的那一行）。**先讀最後一行**。典型的一個例子：

```text
KeyError: 'Disease'
```

`KeyError` 的意思是「*我找了這個鍵（欄位名稱），但它不存在*」。欄位叫做 `disease`，d 是小寫：Python 區分大小寫。

下面的儲存格故意犯這個錯誤。`try:` / `except` 這幾行會接住錯誤並印出來，讓筆記本能繼續執行。

**請 Tala 解釋一行程式碼。** 當某一行讓你困惑時，只貼**那一行**（程式碼，絕不貼資料），用平常的話發問。例如：

> *請對初學者逐字解釋這一行 pandas，並說明在一個只有 3 列虛構資料的小表格上，它會印出什麼：`df["disease"].value_counts()`。*

然後把這個解釋，和那一行真正印出的結果對照檢查。

````python
try:
    df["Disease"]                 # 大寫 D：沒有這個欄位
except KeyError as error:
    print("KeyError:", error)
    print("真正存在的欄位，開頭是：", list(df.columns)[:4])
````
