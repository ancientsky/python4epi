## 1. Mga value at variable

Ang **variable** ay pangalan ng isang value — parang pagsulat ng label sa folder. Ang
ibig sabihin ng `=` ay *"ilagay ang value na ito sa folder na iyan"*, hindi *"ay katumbas
ng"*.

- Ang `7` ay buong numero (*integer*).
- Ang `"Ilog"` ay text (*string*): laging nasa loob ng quotes ang text.
- Ang `True` / `False` ay oo/hindi na value (*booleans*).

Ipinapakita ng `print()` ang value sa screen. Ang text pagkatapos ng `#` ay **comment**:
hindi ito pinapansin ng Python; tala ito para sa tao. Naglalagay ang **f-string** ng mga
value sa loob ng pangungusap: isulat ang `f` bago ang quotes at ang pangalan ng variable sa
loob ng `{ }` (huling linya sa ibaba).

````python
forms_on_desk = 6            # ilang papel na form ang hawak ni Maya ngayong umaga
barangay = "Ilog"            # nasa loob ng quotes ang text
flood_reached_house = True   # oo/hindi na value

print(forms_on_desk)
print(barangay)
print(flood_reached_house)
print(f"May {forms_on_desk} form si Maya mula sa Barangay {barangay}.")
````

## 2. Lists: maraming value sa iisang variable

Nakaayos ang mga value sa isang **list**, sa loob ng square brackets `[ ]`. Nagsisimula sa
**0** ang bilang ng Python sa mga posisyon, kaya ang *unang* item ay `ages[0]` — ang
pinakakaraniwang ikinagugulat ng mga baguhan (at karaniwang mali rin ng AI).

Binibilang ng `len()` ang mga item; pinagsasama-sama ng `sum()` ang mga ito.

````python
ages = [34, 9, 41, 15, 52, 23]   # edad sa anim na form (gawa-gawa)

print(len(ages))     # ilan ang item?
print(ages[0])       # ang UNANG item ay nasa posisyon 0
print(ages[-1])      # ang ibig sabihin ng -1 ay "ang huling item"
print(sum(ages) / len(ages))   # ang mean na edad
````

## 3. Dictionaries: isang form, maraming field

Nag-iimbak ang **dictionary** ng mga value na *may label*: isang **key** (ang label) at
isang **value**, sa loob ng curly braces `{ }`. Kasyang-kasya ang isang notification form
sa iisang dictionary. Letra ang ginagamit natin, hindi pangalan, para makilala ang bawat
form — magandang gawi sa privacy mula sa unang araw.

Ang buong patong ng form ni Maya ay isang **list ng dictionaries**, isa bawat form.
Tingnang mabuti ang anim na form: bibilangin mo sila nang mano-mano mamaya.

````python
form_a = {"form": "A", "barangay": "Ilog", "disease": "leptospirosis", "age": 34, "waded": True}

print(form_a["disease"])   # basahin ang isang value gamit ang key nito
print(form_a["age"])

# Ang buong patong ni Maya: isang LIST ng dictionaries, isa bawat form (lahat imbento)
forms = [
    {"form": "A", "barangay": "Ilog",   "disease": "leptospirosis", "age": 34, "waded": True},
    {"form": "B", "barangay": "Ilog",   "disease": "dengue",        "age": 9,  "waded": False},
    {"form": "C", "barangay": "Wawa",   "disease": "leptospirosis", "age": 41, "waded": True},
    {"form": "D", "barangay": "Sapa",   "disease": "dengue",        "age": 15, "waded": False},
    {"form": "E", "barangay": "Ilog",   "disease": "leptospirosis", "age": 52, "waded": False},
    {"form": "F", "barangay": "Talaba", "disease": "dengue",        "age": 23, "waded": True},
]
print(f"{len(forms)} form")
````

## 4. Isang loop at isang `if`: pagbibilang na parang tally sheet

Tinanong ni Maya ang AI assistant niyang si Tala:

> *May Python list akong `forms`. Bawat item ay dictionary na may keys na `form`,
> `barangay`, `disease`, `age`, `waded` (True/False). Dalawang gawa-gawang item:
> `{"form": "A", "barangay": "Ilog", "disease": "leptospirosis", "age": 34, "waded": True}` at
> `{"form": "B", "barangay": "Ilog", "disease": "dengue", "age": 9, "waded": False}`.
> Sumulat ng code na madaling intindihin ng baguhan na bibilang ng mga leptospirosis
> form. Ipaliwanag ang bawat linya.*

**Bilangin muna nang mano-mano.** Isa-isahin ang listahan: ang mga leptospirosis form ay
A, C at E — **3**. Isulat iyan *bago* ka magpatakbo ng kahit ano.

Nasa ibaba ang sagot ni Tala. Basahin ito na parang tally sheet: magsimula sa zero;
tingnan ang bawat form; **kung** leptospirosis ito, magdagdag ng isa. Ang mga linyang
naka-indent (4 na space) ay bahagi ng loop o ng `if`. Ang huling linya ang ating
**check**: titigil nang may error ang `assert` kapag mali ang pahayag at tahimik lang
kapag tama — magandang balita ang katahimikan.

````python
lepto_count = 0                      # simulan ang tally sa zero
for f in forms:                      # tingnan ang bawat form, isa-isa
    if f["disease"] == "leptospirosis":   # nagtatanong ang == ng "pareho ba?"
        lepto_count = lepto_count + 1     # magdagdag ng isang tally mark

print(f"Mga leptospirosis form: {lepto_count}")
assert lepto_count == 3, "Hindi tugma ang tally sa hand count!"   # ang ating check
````

## 5. Nagkamali si Tala: ang denominator

Sunod na tanong ni Maya: *"Ilang porsiyento ng mga **leptospirosis** form ang lumusong sa
baha?"* Ipinadala ni Tala ang code sa susunod na cell. Tumatakbo ito nang walang kahit
anong error. Bago mo ito patakbuhin, bilangin nang mano-mano: ilan ang leptospirosis form,
at ilan sa kanila ang lumusong? (Ang ibig sabihin ng `and` ay dapat parehong totoo ang
dalawang test.)

````python
# Bersiyon ni Tala, pinatakbo nang walang binago (may mali ito -- nakikita mo ba bago magbasa pa?)
waded_lepto = 0
for f in forms:
    if f["disease"] == "leptospirosis" and f["waded"]:
        waded_lepto = waded_lepto + 1
percent_waded = waded_lepto / len(forms) * 100
print(f"Sagot ni Tala: {percent_waded:.1f}%")
````

**Bilangin nang mano-mano.** Mga leptospirosis form: A, C, E (3). Sa mga iyon, ang
lumusong: A at C (2). Kaya 2 sa 3 ang sagot — mga **66.7 %**, hindi 33.3 %.

Tama ang **numerator** (2). Mali ang **denominator**: binibilang ng `len(forms)` ang
*lahat ng anim* na form, kasama ang dengue. Kasinghusay lang ng denominator nito ang isang
porsiyento — laging itanong ang *"porsiyento ng ano?"*

Ang ayos: sumulat ng maliit na **function** — isang recipe na may pangalan at puwedeng
gamitin ulit. Sinisimulan ito ng `def`, ang mga value sa loob ng bracket ang mga input
nito, at ibinabalik ng `return` ang resulta.

Ang linyang `[f for f in forms if f["disease"] == "leptospirosis"]` ay isang-linyang loop:
gumagawa ito ng bagong list na ang mga form lang na pumasa sa test ang itinitira. Basahin
ito bilang *"bawat form f, mula sa forms, kung leptospirosis ito"*. Ang mga label na
`: int` at `-> float` ay opsiyonal na tala kung anong klase ng value ang pumapasok at
lumalabas. Tumatanggi rin ang function na mag-divide sa zero nang may malinaw na mensahe
— at sa dulo ng cell, **chine-check** ang hand count at ang edge case na iyon.

````python
def percent(part: int, whole: int) -> float:
    # Ibalik ang part bilang porsiyento ng whole.
    if whole == 0:
        raise ValueError("Ang denominator ay 0: walang puwedeng kunan ng porsiyento.")
    return part / whole * 100


lepto_forms = [f for f in forms if f["disease"] == "leptospirosis"]   # lepto forms lang
waded_lepto = len([f for f in lepto_forms if f["waded"]])

answer = percent(waded_lepto, len(lepto_forms))
print(f"{waded_lepto} sa {len(lepto_forms)} leptospirosis form ang lumusong: {answer:.1f}%")

# Check 1: tugma sa hand count.  Check 2: ang edge case ng zero na denominator.
assert round(answer, 1) == 66.7
try:
    percent(1, 0)
except ValueError as error:
    print("Mabuti, tumatanggi ang function:", error)
````

## 6. Mula anim na form hanggang sa buong line list: pandas

Ayos lang na bilangin ang anim na form gamit ang loop. Para sa daan-daang row, gumagamit
tayo ng **pandas**, ang spreadsheet ng Python. Nilo-load ito ng `import pandas as pd` at
binibigyan ng maikling pangalang `pd`.

Bubuksan natin ang **malinis at de-identified** na line list ng Santa Luz (walang
pangalan, telepono o address — ipinapakita ng kabanata 02 kung paano ito naging ganoon).
Relative sa folder ng repository ang path. Ibinibigay ng `.shape` ang (rows, columns);
inililista ng `df.columns` ang pangalan ng mga column — kopyahin ang mga ito nang
**eksakto** kapag humihingi ka ng code kay Tala (`date_onset`, hindi "onset date");
ipinapakita ng `.head()` ang unang limang row.

````python
import pandas as pd

df = pd.read_csv("data/synthetic/ph/santa_luz_linelist.csv")
print(df.shape)
print(list(df.columns))
df.head()
````

Ginagawa ng `value_counts()` ang ginawa ng loop natin — isang tally — para sa buong
column nang sabay. Saka natin ito **iche-check**: dapat tumugma ang kabuuan sa bilang ng
rows.

Saka ang unang chart: mga kaso ayon sa **band** ng barangay (riverside = tabing-ilog,
coastal = baybayin, upland = kabundukan) at sakit. Gumagawa ang `pd.crosstab` ng maliit na
table ng bilang; iginuguhit ito ng `.plot.barh` bilang pahalang na bars.

````python
counts = df["disease"].value_counts()
print(counts)

assert counts.sum() == len(df), "May rows na hindi nabilang!"
print(f"Kabuuan: {counts.sum()} kaso = {len(df)} rows")

import matplotlib.pyplot as plt

table = pd.crosstab(df["band"], df["disease"]).loc[["upland", "coastal", "riverside"]]
print(table)

ax = table.plot.barh(color=["#CE1126", "#0038A8"], figsize=(7, 3.5))
ax.set_xlabel("Bilang ng kaso (naiulat hanggang 30 Set 2026)")
ax.set_ylabel("")
ax.set_title("Santa Luz City (kathang-isip): mga kaso ayon sa band at sakit")
ax.legend(title="")
plt.tight_layout()
plt.show()
````

## 7. Basahin nang kalmado ang error

Mensahe ang error, hindi insulto. Binabasa ng Python ang code mo, tumatama sa problema at
sinasabi sa iyo **kung ano** ang mali (ang error *type*) at **saan** (ang huling linyang
sinubukan nito). Basahin muna ang **huling linya**. Isang karaniwang halimbawa:

```text
KeyError: 'Disease'
```

Ang ibig sabihin ng `KeyError` ay *"Hinanap ko ang key na ito (pangalan ng column) at
wala ito"*. `disease` ang pangalan ng column, maliit na *d*: case-sensitive ang Python.

Sinasadya ng cell sa ibaba ang pagkakamaling iyan. Hinuhuli ng mga linyang `try:` /
`except` ang error at ipini-print ito, kaya tuloy pa rin ang notebook.

**Pagpapaliwanag kay Tala ng isang linya.** Kapag nalilito ka sa isang linya, i-paste ang
**linyang iyon lang** (code, hindi kailanman datos) at magtanong sa simpleng salita.
Halimbawa:

> *Ipaliwanag mo ang linyang ito ng pandas sa isang baguhan, salita por salita, at sabihin
> kung ano ang ipi-print nito para sa maliit na gawa-gawang table na may 3 row:
> `df["disease"].value_counts()`.*

Pagkatapos, i-check ang paliwanag laban sa talagang ipini-print ng linya.

````python
try:
    df["Disease"]                 # malaking D: walang ganitong column
except KeyError as error:
    print("KeyError:", error)
    print("Ang mga column na mayroon ay nagsisimula sa:", list(df.columns)[:4])
````
