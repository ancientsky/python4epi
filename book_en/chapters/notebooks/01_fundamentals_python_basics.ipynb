松柏護理之家爆發 *Legionella pneumophila* 群聚事件，280 位住民中有 121 人感染、19 人死亡。

在你學會用 pandas 讀 CSV 之前，先用最基本的 Python 把關鍵指標算出來。

```python
# Google Colab setup -- 若在本機執行可跳過此 cell
# 這個 cell 只有在 Google Colab 上才會執行
# 它做的事：把教材原始碼從 GitHub 下載到 Colab，然後安裝需要的套件
import sys
import os
if 'google.colab' in sys.modules:   # 偵測是否在 Colab 環境中
    !git clone https://github.com/ancientsky/python4epi.git /content/python4epi 2>/dev/null || true
    os.chdir('/content/python4epi')  # 切換到教材目錄
    !pip install -q -e .             # 安裝教材的 Python 套件
```

```python
# --- 1) 數值變數：把群聚事件的數據存起來 ---
# 在 Python 裡，# 後面的文字是「註解」，Python 不會執行
# 註解是寫給人看的筆記，幫助你（和同事）理解程式碼的用途

total_residents = 280       # 住民總數
infected = 121              # 感染人數（含無症狀）
confirmed = 89              # 確診個案
probable = 25               # 可能個案
hospitalized = 68           # 住院人數
icu = 23                    # ICU 人數
deaths = 19                 # 死亡人數

# --- 2) 計算核心指標 ---
# 除法 / 得到的是浮點數（有小數點的數字）
attack_rate = infected / total_residents   # 侵襲率 = 感染人數 ÷ 全體住民
cfr = deaths / infected                    # 致死率 = 死亡人數 ÷ 感染人數
hosp_rate = hospitalized / infected        # 住院率 = 住院人數 ÷ 感染人數

# f-string 格式化輸出：f"..." 裡面的 {變數} 會被替換成值
# :.2% → 乘以 100 並顯示到小數點後 2 位，加上 % 符號
print(f"侵襲率 (Attack rate): {attack_rate:.2%}")
print(f"致死率 (CFR): {cfr:.2%}")
print(f"住院率: {hosp_rate:.2%}")
print(f"ICU 率（住院者中）: {icu / hospitalized:.2%}")
```

```python
# --- 3) 字典（dict）：把同一主題的資料組織在一起 ---
# 字典用大括號 {} 包住，每組是 "key": value
# 就像一個有標籤的收納盒，用 key 可以快速找到對應的 value
outbreak = {
    "facility": "松柏護理之家",           # key = "facility", value = "松柏護理之家"
    "pathogen": "Legionella pneumophila",  # key = "pathogen", value = "Legionella..."
    "total_residents": 280,
    "infected": 121,
    "confirmed": 89,
    "deaths": 19,
    "hospitalized": 68,
}

# 用方括號 + key 取值：outbreak["deaths"] → 19
cfr = outbreak["deaths"] / outbreak["infected"]
print(f"{outbreak['facility']} — {outbreak['pathogen']}")
print(f"  確診: {outbreak['confirmed']}  可能: {outbreak['infected'] - outbreak['confirmed'] - 7}")
print(f"  CFR: {cfr:.2%}")
```

```python
# --- 4) 列表（list）：存放一組同類資料 ---
# 列表用方括號 [] 包住，每個元素用逗號隔開
# 就像一排格子櫃，每格放一個值，從 0 號開始編號
floor_wing_names = ["1A", "1B", "2A", "2B", "3A", "3B"]
floor_wing_cases = [15, 10, 24, 25, 20, 27]    # 各翼區感染人數
floor_wing_total = [44, 47, 44, 50, 48, 47]    # 各翼區住民總數

# 用 for 迴圈 + range 遍歷每個翼區
# range(len(...)) 產生 0, 1, 2, 3, 4, 5 六個數字
print("=== 各翼區侵襲率 ===")
for i in range(len(floor_wing_names)):
    ar = floor_wing_cases[i] / floor_wing_total[i]  # 侵襲率 = 感染數 ÷ 住民數
    print(f"  {floor_wing_names[i]}: {ar:.1%}  ({floor_wing_cases[i]}/{floor_wing_total[i]})")
```

```python
# --- 5) 條件判斷（if / elif / else）---
# 根據條件走不同的分支，就像馬路上的岔路口
# if 條件成立 → 執行這段；elif 是「否則如果」；else 是「都不是的話」
cfr = deaths / infected

if cfr > 0.15:   # 致死率 > 15%
    print(f"⚠ 致死率 {cfr:.1%}（>15%）：建議升級應變層級、通知醫院準備床位")
elif cfr > 0.10:  # 10% < 致死率 ≤ 15%
    print(f"致死率 {cfr:.1%}（10-15%）：持續加強監測")
else:             # 致死率 ≤ 10%
    print(f"致死率 {cfr:.1%}（<10%）：維持常規應變")
```

```python
# --- 6) 函式（function）：把計算邏輯包成可重用的工具 ---
# def 函式名稱(參數):  ← 定義函式
#     return 結果      ← 回傳計算結果
# 定義好之後，用 函式名稱(值) 來呼叫

def calc_attack_rate(cases, population):
    """計算侵襲率，回傳浮點數。"""  # ← 這是 docstring（文件字串），說明函式用途
    if population == 0:
        raise ValueError("分母（population）不能為 0")  # 防止除以 0
    return cases / population

# 呼叫函式：把數字傳進去，得到侵襲率
print(f"全機構侵襲率: {calc_attack_rate(121, 280):.2%}")

# 同一個函式可以用在不同的翼區——這就是函式的威力
print(f"3 樓 B 翼侵襲率: {calc_attack_rate(27, 47):.2%}")   # 侵襲率最高
print(f"1 樓 B 翼侵襲率: {calc_attack_rate(10, 47):.2%}")   # 侵襲率最低
```
