# 01 Python Basics: Learning From a Legionnaires' Disease Outbreak

An outbreak of *Legionella pneumophila* has broken out at Songbai Nursing Home: of the 280 residents, 121 are infected and 19 have died.

Before you learn to read a CSV with pandas, let's compute the key metrics using the most basic Python.

```python
# Google Colab setup -- skip this cell if running locally
# This cell only runs on Google Colab
# What it does: download the course source from GitHub into Colab, then install the required packages
import sys
import os
if 'google.colab' in sys.modules:   # Detect whether we're in the Colab environment
    !git clone https://github.com/ancientsky/python4epi.git /content/python4epi 2>/dev/null || true
    os.chdir('/content/python4epi')  # Switch to the course directory
    !pip install -q -e .             # Install the course's Python package
```

```python
# --- 1) Numeric variables: store the outbreak data ---
# In Python, text after # is a "comment"; Python does not run it
# Comments are notes for humans, helping you (and your colleagues) understand what the code is for

total_residents = 280       # Total number of residents
infected = 121              # Number infected (including asymptomatic)
confirmed = 89              # Confirmed cases
probable = 25               # Probable cases
hospitalized = 68           # Number hospitalized
icu = 23                    # Number in ICU
deaths = 19                 # Number of deaths

# --- 2) Compute the core metrics ---
# Division / gives a float (a number with a decimal point)
attack_rate = infected / total_residents   # Attack rate = number infected / total residents
cfr = deaths / infected                    # CFR = number of deaths / number infected
hosp_rate = hospitalized / infected        # Hospitalization rate = number hospitalized / number infected

# f-string formatted output: inside f"...", each {variable} is replaced by its value
# :.2% → multiply by 100 and show 2 decimal places, with a % sign
print(f"Attack rate: {attack_rate:.2%}")
print(f"CFR: {cfr:.2%}")
print(f"Hospitalization rate: {hosp_rate:.2%}")
print(f"ICU rate (among hospitalized): {icu / hospitalized:.2%}")
```

```python
# --- 3) Dictionaries (dict): organize data on the same topic together ---
# A dictionary is wrapped in curly braces {}, each pair being "key": value
# It's like a labeled storage box: the key lets you quickly find the matching value
outbreak = {
    "facility": "Songbai Nursing Home",    # key = "facility", value = "Songbai Nursing Home"
    "pathogen": "Legionella pneumophila",  # key = "pathogen", value = "Legionella..."
    "total_residents": 280,
    "infected": 121,
    "confirmed": 89,
    "deaths": 19,
    "hospitalized": 68,
}

# Retrieve a value with square brackets + key: outbreak["deaths"] → 19
cfr = outbreak["deaths"] / outbreak["infected"]
print(f"{outbreak['facility']} — {outbreak['pathogen']}")
print(f"  Confirmed: {outbreak['confirmed']}  Probable: {outbreak['infected'] - outbreak['confirmed'] - 7}")
print(f"  CFR: {cfr:.2%}")
```

```python
# --- 4) Lists (list): store a group of same-kind data ---
# A list is wrapped in square brackets [], with each element separated by a comma
# It's like a row of pigeonholes: each holds one value, numbered starting from 0
floor_wing_names = ["1A", "1B", "2A", "2B", "3A", "3B"]
floor_wing_cases = [15, 10, 24, 25, 20, 27]    # Number infected in each wing
floor_wing_total = [44, 47, 44, 50, 48, 47]    # Total residents in each wing

# Use a for loop + range to iterate over each wing
# range(len(...)) produces the six numbers 0, 1, 2, 3, 4, 5
print("=== Attack rate by wing ===")
for i in range(len(floor_wing_names)):
    ar = floor_wing_cases[i] / floor_wing_total[i]  # Attack rate = number infected / number of residents
    print(f"  {floor_wing_names[i]}: {ar:.1%}  ({floor_wing_cases[i]}/{floor_wing_total[i]})")
```

```python
# --- 5) Conditionals (if / elif / else) ---
# Take different branches depending on the condition, like a fork in the road
# if the condition holds → run this block; elif means "else if"; else means "if none of the above"
cfr = deaths / infected

if cfr > 0.15:   # CFR > 15%
    print(f"⚠ CFR {cfr:.1%} (>15%): recommend escalating the response level and asking hospitals to prepare beds")
elif cfr > 0.10:  # 10% < CFR ≤ 15%
    print(f"CFR {cfr:.1%} (10-15%): keep strengthening surveillance")
else:             # CFR ≤ 10%
    print(f"CFR {cfr:.1%} (<10%): maintain routine response")
```

```python
# --- 6) Functions: package your calculation logic into a reusable tool ---
# def function_name(parameters):  ← define the function
#     return result               ← return the computed result
# Once defined, call it with function_name(value)

def calc_attack_rate(cases, population):
    """Calculate the attack rate; returns a float."""  # ← this is a docstring, describing what the function does
    if population == 0:
        raise ValueError("The denominator (population) cannot be 0")  # Prevent division by 0
    return cases / population

# Call the function: pass in numbers, get back the attack rate
print(f"Facility-wide attack rate: {calc_attack_rate(121, 280):.2%}")

# The same function works for different wings — that's the power of functions
print(f"3rd floor wing B attack rate: {calc_attack_rate(27, 47):.2%}")   # Highest attack rate
print(f"1st floor wing B attack rate: {calc_attack_rate(10, 47):.2%}")   # Lowest attack rate
```
