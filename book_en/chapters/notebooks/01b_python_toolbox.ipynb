## Part 1：Python 語法規範

```python
# --- 1) 縮排：冒號後面的下一行要空 4 格 ---
# Python 用「縮排」（indentation）來表示程式碼的層級關係
# 規則很簡單：看到冒號 : → 下一行往右縮 4 個空格
# 如果忘了縮排，Python 會噴 IndentationError

cfr = 19 / 121   # 致死率 = 死亡 19 / 感染 121

# ✅ 正確縮排：if 後面的 print 都往右縮 4 格
if cfr > 0.15:                              # 條件：致死率 > 15%？
    print("致死率偏高，建議升級應變層級")      # 縮 4 格 → 屬於 if 的區塊
    print("請通知指揮中心")                   # 同樣縮 4 格 → 也屬於 if 的區塊

# 巢狀縮排：for 迴圈裡面又有 if，就要縮兩層（4+4=8 格）
floor_cases = [15, 10, 24, 25, 20, 27]      # 6 個翼區的感染人數
for cases in floor_cases:                    # 逐一取出每個翼區的感染數
    if cases > 20:                           # 第一層縮排（4 格）：for 的區塊
        print(f"{cases} 人感染，需要重點關注")  # 第二層縮排（8 格）：if 的區塊
```

```python
# --- 2) import：借用別人的工具 ---
# Python 內建很多「模組」（module），就像工具箱裡的不同抽屜
# 用 import 把需要的工具借來用，有三種寫法：

# 寫法 1：import 整個模組 → 用「模組名.功能名」呼叫
import math                          # 借來整個 math（數學）模組
print(f"sqrt(121) = {math.sqrt(121)}")  # math.sqrt() = 開根號

# 寫法 2：只借一個功能 → 直接用功能名，不用加模組名
from math import sqrt                # 只從 math 借 sqrt 這個功能
print(f"sqrt(121) = {sqrt(121)}")    # 直接寫 sqrt()，不用 math.sqrt()

# 寫法 3：取別名（alias）→ 用短名稱代替長名稱
import statistics as stats           # statistics 太長，取別名叫 stats
# 之後 Ch02 你會看到 import pandas as pd — 同樣道理

# 用 statistics 模組分析 10 位感染者的年齡
ages = [72, 68, 81, 75, 90, 66, 78, 85, 73, 69]  # 10 位感染者的年齡列表
print(f"平均年齡: {stats.mean(ages):.1f}")    # stats.mean() = 算平均值
print(f"中位數:   {stats.median(ages):.1f}")  # stats.median() = 算中位數
print(f"標準差:   {stats.stdev(ages):.1f}")   # stats.stdev() = 算標準差
```

```python
# --- 3) 型別與轉換 ---

# type() 檢查型別
print(type(280))           # int
print(type(43.2))          # float
print(type("confirmed"))   # str
print(type(True))          # bool

# 型別轉換：CSV 讀進來的數字常常是文字
infected_str = "121"
total_str = "280"
attack_rate = int(infected_str) / int(total_str)
print(f"\n侵襲率: {attack_rate:.2%}")

# 布林值與邏輯運算
cfr = 19 / 121
high_cfr = cfr > 0.10
many_cases = 121 > 100
print(f"\n高致死率? {high_cfr}")
print(f"大量感染? {many_cases}")
print(f"兩者皆是? {high_cfr and many_cases}")
```

## Part 2：Debug 生存技能

```python
# --- 4) 5 大常見錯誤示範（每種都修好） ---

# 1) NameError：變數名拼錯
infected = 121
# infceted  # 取消註解會看到 NameError
print(f"NameError 修正：infected = {infected}")

# 2) TypeError：文字做除法
# "121" / "280"  # 取消註解會看到 TypeError
print(f"TypeError 修正：{int('121') / int('280'):.2%}")

# 3) KeyError：字典 key 不存在
outbreak = {"deaths": 19, "infected": 121}
# outbreak["death"]  # 取消註解會看到 KeyError
print(f"KeyError 修正：{outbreak['deaths']}")

# 4) IndexError：索引超出範圍
wings = ["1A", "1B", "2A", "2B", "3A", "3B"]
# wings[6]  # 取消註解會看到 IndexError
print(f"IndexError 修正：{wings[5]}（最後一個元素用 index 5）")

# 5) 查看所有可用 key
print(f"\n字典的所有 key: {list(outbreak.keys())}")
```

```python
# --- 5) try/except：優雅處理意外 ---

# 模擬從 CSV 讀到的年齡欄位（有遺漏值）
ages_raw = ["72", "68", "N/A", "75", "unknown", "66", "81", "90"]

ages = []
skipped = 0
for val in ages_raw:
    try:
        ages.append(int(val))
    except ValueError:
        skipped += 1

print(f"成功轉換 {len(ages)} 筆，跳過 {skipped} 筆")
print(f"平均年齡: {sum(ages) / len(ages):.1f}")
```

## Part 3：實用開發技巧

```python
# --- 6) 字串方法與迴圈進階 ---

# 字串清理
raw_name = "  Legionella pneumophila  "
print(f"strip: '{raw_name.strip()}'")

raw_severity = "Mild,Moderate,Severe"
levels = raw_severity.split(",")
print(f"split: {levels}")

# in 成員檢查
pathogen = "Legionella pneumophila serogroup 1"
print(f"\nstartswith: {pathogen.startswith('Legionella')}")
print(f"in: {'serogroup' in pathogen}")

# enumerate() — 同時拿到索引和值
daily_cases = [3, 7, 12, 8, 15]
print("\n每日新增病例：")
for i, cases in enumerate(daily_cases, start=1):
    print(f"  Day {i}: {cases} 例")
```
