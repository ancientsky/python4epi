## Part 1: Python Syntax Rules

```python
# --- 1) Indentation: the line after a colon must be indented 4 spaces ---
# Python uses "indentation" to express the hierarchy of the code
# The rule is simple: see a colon : → indent the next line 4 spaces to the right
# If you forget to indent, Python throws an IndentationError

cfr = 19 / 121   # case fatality rate = 19 deaths / 121 infected

# ✅ Correct indentation: the prints after if are all indented 4 spaces
if cfr > 0.15:                              # condition: is the CFR > 15%?
    print("CFR is high; recommend escalating the response level")      # indented 4 spaces -> part of the if block
    print("Please notify the command center")                   # also indented 4 spaces -> also part of the if block

# Nested indentation: a for loop with an if inside needs two levels (4+4=8 spaces)
floor_cases = [15, 10, 24, 25, 20, 27]      # infected counts for the 6 wings
for cases in floor_cases:                    # take each wing's infected count in turn
    if cases > 20:                           # first indent level (4 spaces): the for block
        print(f"{cases} infected, needs close attention")  # second indent level (8 spaces): the if block
```

```python
# --- 2) import: borrowing other people's tools ---
# Python has many built-in "modules"—like different drawers in a toolbox
# Use import to borrow the tools you need; there are three forms:

# Form 1: import the whole module → call it as "module_name.feature_name"
import math                          # borrow the whole math module
print(f"sqrt(121) = {math.sqrt(121)}")  # math.sqrt() = square root

# Form 2: borrow just one feature → use the feature name directly, no module prefix
from math import sqrt                # borrow only sqrt from math
print(f"sqrt(121) = {sqrt(121)}")    # write sqrt() directly, no math.sqrt()

# Form 3: give it an alias → use a short name in place of a long one
import statistics as stats           # statistics is long, so alias it as stats
# later in Ch02 you'll see import pandas as pd — same idea

# use the statistics module to analyze the ages of 10 infected people
ages = [72, 68, 81, 75, 90, 66, 78, 85, 73, 69]  # list of the 10 infected people's ages
print(f"Mean age: {stats.mean(ages):.1f}")    # stats.mean() = compute the mean
print(f"Median:   {stats.median(ages):.1f}")  # stats.median() = compute the median
print(f"Std dev:  {stats.stdev(ages):.1f}")   # stats.stdev() = compute the standard deviation
```

```python
# --- 3) Types and conversion ---

# type() checks the type
print(type(280))           # int
print(type(43.2))          # float
print(type("confirmed"))   # str
print(type(True))          # bool

# type conversion: numbers read in from a CSV are often text
infected_str = "121"
total_str = "280"
attack_rate = int(infected_str) / int(total_str)
print(f"\nAttack rate: {attack_rate:.2%}")

# booleans and logical operations
cfr = 19 / 121
high_cfr = cfr > 0.10
many_cases = 121 > 100
print(f"\nHigh CFR? {high_cfr}")
print(f"Many infections? {many_cases}")
print(f"Both true? {high_cfr and many_cases}")
```

## Part 2: Debugging Survival Skills

```python
# --- 4) The 5 common errors demonstrated (each one fixed) ---

# 1) NameError: variable name typo
infected = 121
# infceted  # uncomment to see a NameError
print(f"NameError fixed: infected = {infected}")

# 2) TypeError: dividing text
# "121" / "280"  # uncomment to see a TypeError
print(f"TypeError fixed: {int('121') / int('280'):.2%}")

# 3) KeyError: the dictionary key doesn't exist
outbreak = {"deaths": 19, "infected": 121}
# outbreak["death"]  # uncomment to see a KeyError
print(f"KeyError fixed: {outbreak['deaths']}")

# 4) IndexError: index out of range
wings = ["1A", "1B", "2A", "2B", "3A", "3B"]
# wings[6]  # uncomment to see an IndexError
print(f"IndexError fixed: {wings[5]} (the last element uses index 5)")

# 5) view all available keys
print(f"\nAll dictionary keys: {list(outbreak.keys())}")
```

```python
# --- 5) try/except: handling the unexpected gracefully ---

# simulate an age column read from a CSV (with missing values)
ages_raw = ["72", "68", "N/A", "75", "unknown", "66", "81", "90"]

ages = []
skipped = 0
for val in ages_raw:
    try:
        ages.append(int(val))
    except ValueError:
        skipped += 1

print(f"Converted {len(ages)} successfully, skipped {skipped}")
print(f"Mean age: {sum(ages) / len(ages):.1f}")
```

## Part 3: Practical Development Techniques

```python
# --- 6) Advanced string methods and loops ---

# string cleaning
raw_name = "  Legionella pneumophila  "
print(f"strip: '{raw_name.strip()}'")

raw_severity = "Mild,Moderate,Severe"
levels = raw_severity.split(",")
print(f"split: {levels}")

# in membership check
pathogen = "Legionella pneumophila serogroup 1"
print(f"\nstartswith: {pathogen.startswith('Legionella')}")
print(f"in: {'serogroup' in pathogen}")

# enumerate() — get the index and value at once
daily_cases = [3, 7, 12, 8, 15]
print("\nDaily new cases:")
for i, cases in enumerate(daily_cases, start=1):
    print(f"  Day {i}: {cases} cases")
```
