```python
# === Problem 1: import and statistics ===
import statistics as stats

hospital_days = [5, 3, 12, 7, 4, 9, 6, 8]

mean_days = stats.mean(hospital_days)
median_days = stats.median(hospital_days)

print(f"Mean length of stay: {mean_days:.1f}")
print(f"Median length of stay: {median_days:.1f}")
```

```python
# === Problem 2: type conversion ===
cases_str = "121"
population_str = "280"
deaths_str = "19"

cases = int(cases_str)
population = int(population_str)
deaths = int(deaths_str)

attack_rate = cases / population
cfr = deaths / cases

print(f"Attack rate: {attack_rate:.2%}")
print(f"CFR: {cfr:.2%}")
```

```python
# === Problem 4: try/except ===
wing_data = ["15", "10", "N/A", "25", "missing", "27"]

clean_data = []
skipped = 0
for val in wing_data:
    try:
        clean_data.append(int(val))
    except ValueError:
        skipped += 1

print(f"Success: {len(clean_data)} items")
print(f"Skipped: {skipped} items")
print(f"Sum: {sum(clean_data)}")
```

```python
# === Problem 5: string cleaning + loops ===
exposures_raw = [" Shower ", "shower", "HYDROTHERAPY", " hydrotherapy", "Shower", "unknown"]

cleaned = []
for raw in exposures_raw:
    cleaned.append(raw.strip().lower())

shower_count = cleaned.count("shower")
hydro_count = cleaned.count("hydrotherapy")

print(f"Cleaned: {cleaned}")
print(f"shower appears {shower_count} times")
print(f"hydrotherapy appears {hydro_count} times")
```

```python
# === Problem 6 (challenge): putting it all together ===
patients = [
    {"id": "P001", "floor": "3", "infected": "1"},
    {"id": "P002", "floor": "2", "infected": "0"},
    {"id": "P003", "floor": "3", "infected": "1"},
    {"id": "P004", "floor": "1", "infected": "N/A"},
    {"id": "P005", "floor": "2", "infected": "1"},
    {"id": "P006", "floor": "3", "infected": "0"},
    {"id": "P007", "floor": "1", "infected": "1"},
    {"id": "P008", "floor": "2", "infected": "1"},
]

# tally dictionary: {floor: {"total": N, "infected": N}}
floor_stats = {}

for p in patients:
    floor = p["floor"]
    if floor not in floor_stats:
        floor_stats[floor] = {"total": 0, "infected": 0}
    
    try:
        infected = int(p["infected"])
        floor_stats[floor]["total"] += 1
        floor_stats[floor]["infected"] += infected
    except ValueError:
        # N/A — skip this row and don't count it in the denominator
        pass

print("Attack rate by floor:")
for floor in sorted(floor_stats):
    s = floor_stats[floor]
    ar = s["infected"] / s["total"] if s["total"] > 0 else 0
    print(f"  {floor}F: {ar:.1%} ({s['infected']}/{s['total']})")
```

```python
# === Problem 7: statistics module ===
import statistics
onset_hours = [24, 30, 28, 36, 26, 40, 32, 29, 34, 27]
print(f"Mean = {statistics.mean(onset_hours):.1f} hours")
print(f"Median = {statistics.median(onset_hours):.1f} hours")
print(f"Standard deviation = {statistics.stdev(onset_hours):.1f} hours")
```

```python
# === Problem 8: try/except and type conversion ===
raw = ["12", "8", "N/A", "15", "", "9", "twenty", "20"]
total, valid, bad = 0, 0, 0
for x in raw:
    try:
        total += int(x); valid += 1
    except ValueError:
        bad += 1
print(f"Valid days = {valid}, total cases = {total}, unparseable = {bad}")
```

```python
# === Problem 9 (challenge): defining a function + error handling ===
def cfr(deaths, cases):
    """Case fatality rate = deaths / cases; raises ValueError when cases <= 0."""
    if cases <= 0:
        raise ValueError("cases must be greater than 0")
    return deaths / cases

for d, c in [(12, 180), (5, 0)]:
    try:
        print(f"cfr({d}, {c}) = {cfr(d, c):.1%}")
    except ValueError as e:
        print(f"cfr({d}, {c}) raised an error: {e}")
```
