```python
# === Question 1: Compute the basic metrics for 3rd floor wing B ===
residents_3b = 47
infected_3b = 27
deaths_3b = 5
hospitalized_3b = 18

attack_rate_3b = infected_3b / residents_3b
cfr_3b = deaths_3b / infected_3b
hosp_rate_3b = hospitalized_3b / infected_3b

print(f"Wing 3B attack rate: {attack_rate_3b:.2%}")
print(f"Wing 3B CFR: {cfr_3b:.2%}")
print(f"Wing 3B hospitalization rate: {hosp_rate_3b:.2%}")
```

```python
# === Question 2: Conditionals ===
if attack_rate_3b > 0.50:
    print("Recommend immediately shutting off the shower facilities in this wing")
else:
    print("Continuing to monitor")
```

```python
# === Question 3 (challenge): compare the wings using a dict + loop ===
wings = {
    "1A": {"residents": 44, "infected": 15},
    "1B": {"residents": 47, "infected": 10},
    "2A": {"residents": 44, "infected": 24},
    "2B": {"residents": 50, "infected": 25},
    "3A": {"residents": 48, "infected": 20},
    "3B": {"residents": 47, "infected": 27},
}

max_ar = 0
max_wing = ""

for name, data in wings.items():
    ar = data["infected"] / data["residents"]
    print(f"{name}: attack rate {ar:.1%}  ({data['infected']}/{data['residents']})")
    if ar > max_ar:
        max_ar = ar
        max_wing = name

print(f"\nWing with the highest attack rate: {max_wing} ({max_ar:.1%})")
```

```python
# === Question 4: COVID-19 basic metrics ===
total = 500
cases = 180
deaths = 12
attack_rate = cases / total
cfr = deaths / cases
print(f"Attack rate = {attack_rate:.1%}, CFR = {cfr:.1%}")
```

```python
# === Question 5: dengue fever percentage and rounding ===
reported = 640
severe = 58
severe_pct = round(severe / reported * 100, 1)
print(f"Severe case proportion = {severe_pct}%")
```

```python
# === Question 6: TB monthly reports, list and loop ===
monthly = [42, 38, 45, 50, 61, 55, 48, 40, 39, 44, 52, 47]
total = sum(monthly)
avg = total / len(monthly)
peak_month = monthly.index(max(monthly)) + 1
print(f"Yearly total = {total}, monthly average = {avg:.1f}, peak month = month {peak_month} ({max(monthly)} cases)")
```

```python
# === Question 7: measles vaccination coverage dict ===
coverage = {"忠孝國小": 97.2, "仁愛國小": 88.5, "信義國小": 95.0, "和平國小": 82.3}
lowest = min(coverage, key=coverage.get)
below = [s for s, v in coverage.items() if v < 95]
print(f"Lowest coverage: {lowest} ({coverage[lowest]}%)")
print(f"Schools below the 95% herd immunity threshold ({len(below)}): {below}")
```

```python
# === Question 8 (challenge): influenza incidence rate by district ===
regions = {"北區": (120, 80000), "南區": (60, 95000),
           "東區": (45, 30000), "西區": (30, 70000)}
for name, (cases, pop) in regions.items():
    rate = cases / pop * 100000
    level = "high risk" if rate >= 100 else "general"
    print(f"{name}: {rate:.1f}/100k -> {level}")
```
