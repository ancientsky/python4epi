```python
# === Question 1: Compute the basic metrics for 3rd floor wing B ===
residents_3b = 47
infected_3b = 27
deaths_3b = 5
hospitalized_3b = 18

attack_rate_3b = infected_3b / residents_3b
cfr_3b = deaths_3b / infected_3b
hosp_rate_3b = hospitalized_3b / infected_3b

print(f"Wing 3B attack rate: {attack_rate_3b:.2%}")
print(f"Wing 3B CFR: {cfr_3b:.2%}")
print(f"Wing 3B hospitalization rate: {hosp_rate_3b:.2%}")
```

```python
# === Question 2: Conditionals ===
if attack_rate_3b > 0.50:
    print("Recommend immediately shutting off the shower facilities in this wing")
else:
    print("Continuing to monitor")
```

```python
# === Question 3 (challenge): compare the wings using a dict + loop ===
wings = {
    "1A": {"residents": 44, "infected": 15},
    "1B": {"residents": 47, "infected": 10},
    "2A": {"residents": 44, "infected": 24},
    "2B": {"residents": 50, "infected": 25},
    "3A": {"residents": 48, "infected": 20},
    "3B": {"residents": 47, "infected": 27},
}

max_ar = 0
max_wing = ""

for name, data in wings.items():
    ar = data["infected"] / data["residents"]
    print(f"{name}: attack rate {ar:.1%}  ({data['infected']}/{data['residents']})")
    if ar > max_ar:
        max_ar = ar
        max_wing = name

print(f"\nWing with the highest attack rate: {max_wing} ({max_ar:.1%})")
```
