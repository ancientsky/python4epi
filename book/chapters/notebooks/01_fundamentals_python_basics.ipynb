從零開始：變數、字典、條件，然後計算 attack rate 與 CFR。

```python
cases = 125
population = 2450
attack_rate = cases / population
print(f'Attack rate: {attack_rate:.2%}')
```

```python
report = {'cases': 125, 'deaths': 4, 'population': 2450}
cfr = report['deaths'] / report['cases']
print(f'CFR: {cfr:.2%}')
```

```python
if attack_rate > 0.05:
    print('需要啟動加強監測')
else:
    print('維持常規監測')
```
