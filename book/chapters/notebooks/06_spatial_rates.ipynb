# 06 Spatial Rates
計算每 10 萬人的地區病例率。

```python
import pandas as pd
```

```python
population = pd.DataFrame({'location': ['North','South','East','West'], 'population': [12000,10000,8000,9000]})
cases = pd.read_csv('data/synthetic/line_list.csv').groupby('location').size().rename('cases')
rate = population.merge(cases, on='location', how='left').fillna({'cases': 0})
rate['incidence_per_100k'] = rate['cases'] / rate['population'] * 100000
rate
```
