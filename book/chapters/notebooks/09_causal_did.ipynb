# 09 Causal DiD
Difference-in-Differences 最小示範。

```python
import pandas as pd
import statsmodels.formula.api as smf
```

```python
panel = pd.DataFrame({'week': [1,2,3,4,5,6]*2, 'treated': [0]*6 + [1]*6, 'cases': [30,28,29,27,26,25,31,30,32,24,22,21]})
panel['post'] = (panel['week'] >= 4).astype(int)
model = smf.ols('cases ~ treated + post + treated:post', data=panel).fit()
model.params
```
