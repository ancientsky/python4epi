```python
# === 題目 1：計算 3 樓 B 翼的基本指標 ===
residents_3b = 47
infected_3b = 27
deaths_3b = 5
hospitalized_3b = 18

attack_rate_3b = infected_3b / residents_3b
cfr_3b = deaths_3b / infected_3b
hosp_rate_3b = hospitalized_3b / infected_3b

print(f"3B 翼侵襲率 (Attack rate): {attack_rate_3b:.2%}")
print(f"3B 翼致死率 (CFR): {cfr_3b:.2%}")
print(f"3B 翼住院率: {hosp_rate_3b:.2%}")
```

```python
# === 題目 2：條件判斷 ===
if attack_rate_3b > 0.50:
    print("建議立即關閉該翼區淋浴設備")
else:
    print("持續監測中")
```

```python
# === 題目 3（挑戰題）：用字典 + 迴圈比較各翼區 ===
wings = {
    "1A": {"residents": 44, "infected": 15},
    "1B": {"residents": 47, "infected": 10},
    "2A": {"residents": 44, "infected": 24},
    "2B": {"residents": 50, "infected": 25},
    "3A": {"residents": 48, "infected": 20},
    "3B": {"residents": 47, "infected": 27},
}

max_ar = 0
max_wing = ""

for name, data in wings.items():
    ar = data["infected"] / data["residents"]
    print(f"{name}: 侵襲率 {ar:.1%}  ({data['infected']}/{data['residents']})")
    if ar > max_ar:
        max_ar = ar
        max_wing = name

print(f"\n侵襲率最高的翼區：{max_wing}（{max_ar:.1%}）")
```

```python
# === 題目 4：COVID-19 基本指標 ===
total = 500
cases = 180
deaths = 12
attack_rate = cases / total
cfr = deaths / cases
print(f"侵襲率 = {attack_rate:.1%}、致死率 = {cfr:.1%}")
```

```python
# === 題目 5：登革熱百分比與四捨五入 ===
reported = 640
severe = 58
severe_pct = round(severe / reported * 100, 1)
print(f"重症比例 = {severe_pct}%")
```

```python
# === 題目 6：結核病月報 list 與迴圈 ===
monthly = [42, 38, 45, 50, 61, 55, 48, 40, 39, 44, 52, 47]
total = sum(monthly)
avg = total / len(monthly)
peak_month = monthly.index(max(monthly)) + 1
print(f"全年總數 = {total}，月平均 = {avg:.1f}，最高月份 = {peak_month} 月（{max(monthly)} 例）")
```

```python
# === 題目 7：麻疹疫苗接種率 dict ===
coverage = {"忠孝國小": 97.2, "仁愛國小": 88.5, "信義國小": 95.0, "和平國小": 82.3}
lowest = min(coverage, key=coverage.get)
below = [s for s, v in coverage.items() if v < 95]
print(f"接種率最低：{lowest}（{coverage[lowest]}%）")
print(f"低於群體免疫門檻 95% 的學校（{len(below)} 間）：{below}")
```

```python
# === 題目 8（挑戰題）：流感各區發生率 ===
regions = {"北區": (120, 80000), "南區": (60, 95000),
           "東區": (45, 30000), "西區": (30, 70000)}
for name, (cases, pop) in regions.items():
    rate = cases / pop * 100000
    level = "高風險" if rate >= 100 else "一般"
    print(f"{name}：{rate:.1f}/10萬 → {level}")
```
