```python
import json\nimport pandas as pd\nimport plotly.express as px
```

```python
pop = pd.read_csv('data/synthetic/location_population.csv')\ncases = pd.read_csv('data/synthetic/line_list.csv').groupby('location').size().rename('cases').reset_index()\nrate = pop.merge(cases, on='location', how='left').fillna({'cases': 0})\nrate['incidence_per_100k'] = rate['cases'] / rate['population'] * 100000\nrate
```

```python
with open('data/synthetic/admin_areas.geojson', 'r', encoding='utf-8') as f:\n    geojson = json.load(f)\nfig = px.choropleth(rate, geojson=geojson, locations='location', featureidkey='properties.location', color='incidence_per_100k', color_continuous_scale='Reds', title='Incidence per 100k by Location')\nfig.update_geos(fitbounds='locations', visible=False)\nfig.show()
```
